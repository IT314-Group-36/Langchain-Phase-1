## Config, job state and logging
Caps and limits live here (NFR-12 resilience, NFR-15 cost control). Every log line carries the job's **correlation ID** (NFR-13).

```python

MAX_RETRIES            = 3
MAX_FILE_BYTES         = 300_000
MAX_WALK_FILES         = 20_000
MAX_CHUNKS             = 6_000
MAX_LLM_TOKENS_PER_JOB = 60_000
MANIFEST_MAX_DEPTH     = 4
CHUNK_SIZE, CHUNK_OVERLAP = 1000, 150
EMBED_MODEL            = "sentence-transformers/all-MiniLM-L6-v2"
EMBED_BATCH            = 64
TENANT_ID              = "demo-tenant"

SCHEMA_VERSION = "1.0"


class Job:
    """Per-run state: correlation id, timings, cost counters, exclusions, warnings."""
    def __init__(self):
        self.corr_id  = uuid.uuid4().hex[:8]
        self.t0       = time.time()
        self.timings  = {}
        self.cost     = {"llm_calls": 0, "llm_prompt_tokens": 0, "llm_output_tokens": 0,
                         "embed_chunks": 0, "embed_tokens_est": 0}
        self.excluded = {}
        self.skipped  = Counter()
        self.warnings = []

JOB = None


def log(stage_name, msg):
    cid = JOB.corr_id if JOB else "--------"
    print(f"[{time.strftime('%H:%M:%S')}] [corr={cid}] {stage_name:<11} {msg}", flush=True)


@contextmanager
def stage(name):
    t = time.time()
    log(name, "started")
    try:
        yield
    finally:
        JOB.timings[name] = time.time() - t
        log(name, f"done in {JOB.timings[name]:.1f}s")
```

## S1-8 - Data contracts and schema-enforced LLM output
* **Final contracts** (`TechStack`, `ModuleMap`, `SourceRef`) are what Teams A and C would consume.
* **LLM drafts** are what the model is allowed to propose. The model's output is constrained by a JSON grammar generated from the Pydantic schema, validated, retried a bounded number of times, and then **grounded** against the real repository by plain code (the model proposes, code disposes).

```python

class LanguageShare(BaseModel):
    name: str
    share: float

class TechStack(BaseModel):
    schema_version: str = SCHEMA_VERSION
    languages: List[LanguageShare]
    frameworks: List[str]
    libraries: List[str]
    package_managers: List[str]
    confidence: Literal["high", "medium", "low"]

class SourceRef(BaseModel):
    file_path: str
    start_line: int
    end_line: int

class Module(BaseModel):
    name: str
    summary: str
    paths: List[str]
    sources: List[SourceRef] = Field(min_length=1, max_length=3)

class ModuleDependency(BaseModel):
    model_config = ConfigDict(populate_by_name=True)
    from_: str = Field(alias="from")
    to: str

class ModuleMap(BaseModel):
    schema_version: str = SCHEMA_VERSION
    modules: List[Module]
    entry_points: List[str]
    dependencies: List[ModuleDependency]

class FrameworkGuess(BaseModel):
    name: str
    evidence: str

class StackDraft(BaseModel):
    frameworks: List[FrameworkGuess]
    primary_libraries: List[str]

class ModuleMapDraft(BaseModel):
    modules: List[Module] = Field(min_length=3, max_length=8)
    entry_points: List[str]
    dependencies: List[ModuleDependency]


class SchemaValidationError(Exception):
    """Raised (typed error) when the LLM output is still invalid after MAX_RETRIES."""

class ContextOverflow(Exception):
    pass

_DROP_KEYS = {"title", "description", "default", "examples", "$defs", "definitions", "additionalProperties",
              "minimum", "maximum", "exclusiveMinimum", "exclusiveMaximum",
              "minLength", "maxLength", "format"}

def to_grammar_schema(model_cls) -> dict:
    """Pydantic schema -> plain JSON schema (refs inlined, unsupported keywords removed) for llama.cpp."""
    schema = model_cls.model_json_schema()
    defs = schema.get("$defs", {})

    def resolve(node):
        if isinstance(node, dict):
            if "$ref" in node:
                return resolve(defs[node["$ref"].split("/")[-1]])
            out = {}
            for k, v in node.items():
                if k == "properties":
                    out[k] = {pk: resolve(pv) for pk, pv in v.items()}
                elif k in _DROP_KEYS:
                    continue
                else:
                    out[k] = resolve(v)
            return out
        if isinstance(node, list):
            return [resolve(x) for x in node]
        return node

    return resolve(schema)


def llama3_prompt(system: str, user: str) -> str:
    return ("<|start_header_id|>system<|end_header_id|>\n\n" + system + "<|eot_id|>"
            "<|start_header_id|>user<|end_header_id|>\n\n" + user + "<|eot_id|>"
            "<|start_header_id|>assistant<|end_header_id|>\n\n")


class StructuredChain:
    """prompt template -> LlamaCpp (JSON-grammar constrained) -> Pydantic parser, with bounded retries."""
    def __init__(self, name, system, user, model_cls):
        self.name = name
        self.template = PromptTemplate.from_template(llama3_prompt(system, user))
        self.grammar = LlamaGrammar.from_json_schema(json.dumps(to_grammar_schema(model_cls)))
        self.parser = PydanticOutputParser(pydantic_object=model_cls)

    def __call__(self, **inputs):
        prompt = self.template.format(**inputs)
        n_prompt = llm.get_num_tokens(prompt)
        if n_prompt + MAX_NEW_TOKENS > N_CTX:
            raise ContextOverflow(f"{self.name}: prompt is {n_prompt} tokens, context window is {N_CTX}")
        last_err = None
        for attempt in range(1, MAX_RETRIES + 1):
            if JOB.cost["llm_prompt_tokens"] + n_prompt > MAX_LLM_TOKENS_PER_JOB:
                raise RuntimeError(f"{self.name}: per-job LLM token cap ({MAX_LLM_TOKENS_PER_JOB}) reached")
            captured = {}
            def grab(text):
                captured["n"] = llm.get_num_tokens(text)
                return text
            chain = (self.template
                     | llm.bind(grammar=self.grammar, temperature=0.1 * attempt, stop=["<|eot_id|>"])
                     | RunnableLambda(grab)
                     | self.parser)
            t = time.time()
            JOB.cost["llm_prompt_tokens"] += n_prompt
            try:
                out = chain.invoke(inputs)
            except Exception as e:
                last_err = e
                log("llm", f"{self.name}: attempt {attempt}/{MAX_RETRIES} failed ({type(e).__name__}); retrying")
                continue
            JOB.cost["llm_calls"] += 1
            JOB.cost["llm_output_tokens"] += captured.get("n", 0)
            log("llm", f"{self.name}: ok in {time.time()-t:.1f}s (prompt {n_prompt} tok, output {captured.get('n', 0)} tok)")
            return out
        raise SchemaValidationError(f"{self.name}: output still invalid after {MAX_RETRIES} attempts: {last_err}")
```

## Repository access, file rules and the secret-exclusion guard
All file content is read through **one function** (`safe_read_text`). It refuses sensitive file names, binaries, oversized files and any file whose *content* matches a credential pattern, so nothing excluded can reach chunking, embedding or a prompt (FR-20, DR-03). Excluded files stay visible as nodes in the tree, but their content is never read.

```python

SKIP_DIRS = {".git", "node_modules", "venv", ".venv", "__pycache__", ".idea", ".vscode", "dist", "build",
             "target", "out", ".next", ".nuxt", ".gradle", ".tox", ".mypy_cache", ".pytest_cache",
             "coverage", "site-packages", "vendor", "bower_components", ".terraform"}

SENSITIVE_DIRS = {".ssh", ".aws", ".gnupg"}
SENSITIVE_NAME_GLOBS = [
    ".env", ".env.*", "*.env", "*.pem", "*.key", "*.p12", "*.pfx", "*.jks", "*.keystore", "*.ppk",
    "id_rsa*", "id_dsa*", "id_ecdsa*", "id_ed25519*", "credentials*", "*credentials.json",
    "*secret*", ".npmrc", ".pypirc", ".netrc", ".htpasswd", "service-account*.json", "*.tfvars",
    "terraform.tfstate*", "kubeconfig", "*.kdbx", "*.gpg",
]
SECRET_PATTERNS = [
    re.compile(r"-----BEGIN (?:RSA |EC |DSA |OPENSSH |PGP )?PRIVATE KEY"),
    re.compile(r"AKIA[0-9A-Z]{16}"),
    re.compile(r"gh[pousr]_[A-Za-z0-9]{36,}"),
    re.compile(r"github_pat_[A-Za-z0-9_]{50,}"),
    re.compile(r"xox[baprs]-[A-Za-z0-9-]{10,}"),
    re.compile(r"AIza[0-9A-Za-z_\-]{35}"),
    re.compile(r"sk-[A-Za-z0-9]{32,}"),
    re.compile(r"(?i)\b(?:api[_-]?key|secret|passwd|password|token)\b\s*[:=]\s*['\"][^'\"\s]{16,}['\"]"),
]

EXT_LANG = {
    ".py": "Python", ".js": "JavaScript", ".jsx": "JavaScript", ".mjs": "JavaScript", ".cjs": "JavaScript",
    ".ts": "TypeScript", ".tsx": "TypeScript", ".java": "Java", ".kt": "Kotlin", ".kts": "Kotlin",
    ".go": "Go", ".rs": "Rust", ".rb": "Ruby", ".php": "PHP", ".cs": "C#", ".c": "C", ".h": "C",
    ".cpp": "C++", ".cc": "C++", ".cxx": "C++", ".hpp": "C++", ".swift": "Swift", ".scala": "Scala",
    ".dart": "Dart", ".lua": "Lua", ".sh": "Shell", ".bash": "Shell", ".html": "HTML", ".htm": "HTML",
    ".css": "CSS", ".scss": "SCSS", ".sql": "SQL", ".r": "R", ".ipynb": "Jupyter Notebook",
}
OTHER_LANG = {
    ".md": "Markdown", ".rst": "reStructuredText", ".txt": "Text", ".json": "JSON", ".yaml": "YAML",
    ".yml": "YAML", ".toml": "TOML", ".ini": "INI", ".cfg": "INI", ".conf": "Config", ".xml": "XML",
    ".properties": "Properties", ".gradle": "Gradle",
}
INDEXABLE_EXTS  = (set(EXT_LANG) | set(OTHER_LANG)) - {".ipynb"}
INDEXABLE_NAMES = {"Dockerfile", "Makefile", "Gemfile", "Rakefile", "Procfile", "Pipfile"}
LOCK_FILES = {"package-lock.json", "yarn.lock", "pnpm-lock.yaml", "poetry.lock", "pipfile.lock", "uv.lock",
              "gemfile.lock", "cargo.lock", "composer.lock", "go.sum", "bun.lockb"}

ENTRY_NAMES = {"main.py", "app.py", "manage.py", "wsgi.py", "asgi.py", "__main__.py", "server.py", "run.py",
               "index.js", "index.ts", "main.js", "main.ts", "app.js", "app.ts", "server.js", "server.ts",
               "app.jsx", "app.tsx", "main.jsx", "main.tsx", "index.jsx", "index.tsx", "main.go", "main.rs",
               "lib.rs", "program.cs", "index.php", "application.java", "main.java", "main.kt", "main.c",
               "main.cpp", "index.html"}


def is_sensitive_name(rel: str) -> bool:
    p = PurePosixPath(rel)
    if any(part.lower() in SENSITIVE_DIRS for part in p.parts[:-1]):
        return True
    name = p.name.lower()
    return any(fnmatch.fnmatch(name, g) for g in SENSITIVE_NAME_GLOBS)

def has_secret_content(text: str) -> bool:
    return any(rx.search(text) for rx in SECRET_PATTERNS)

def language_of(rel: str) -> str:
    ext = PurePosixPath(rel).suffix.lower()
    return EXT_LANG.get(ext) or OTHER_LANG.get(ext) or ("Docker" if PurePosixPath(rel).name == "Dockerfile" else "Text")

def is_indexable(rel: str) -> bool:
    p = PurePosixPath(rel)
    low = p.name.lower()
    if low in LOCK_FILES or low.endswith((".min.js", ".min.css", ".map", ".lock")):
        return False
    return p.suffix.lower() in INDEXABLE_EXTS or p.name in INDEXABLE_NAMES or low.startswith("readme")


def safe_read_text(repo_dir: str, rel: str, max_bytes: int = MAX_FILE_BYTES):
    """The ONLY function that reads file content. Returns None for anything excluded."""
    if rel in JOB.excluded:
        return None
    if is_sensitive_name(rel):
        JOB.excluded[rel] = "sensitive file name"
        return None
    full = os.path.join(repo_dir, rel)
    try:
        if os.path.islink(full) or os.path.getsize(full) > max_bytes:
            JOB.skipped["too large / symlink"] += 1
            return None
        with open(full, "rb") as f:
            raw = f.read()
    except OSError:
        JOB.skipped["unreadable"] += 1
        return None
    if b"\x00" in raw[:4096]:
        JOB.skipped["binary"] += 1
        return None
    text = raw.decode("utf-8", errors="replace")
    if has_secret_content(text):
        JOB.excluded[rel] = "credential pattern in content"
        return None
    return text


class RepoError(Exception):
    pass

def parse_repo_url(s: str):
    s = s.strip()
    m = re.match(r"^git@github\.com:([\w.\-]+)/([\w.\-]+?)(?:\.git)?$", s)
    if not m:
        m = re.match(r"^(?:https?://)?(?:www\.)?github\.com/([\w.\-]+)/([\w.\-]+?)(?:\.git)?"
                     r"(?:/tree/([^/\s]+)(?:/\S*)?)?/?$", s)
    if not m:
        raise RepoError("That does not look like a GitHub repository link "
                        "(expected e.g. https://github.com/owner/repo).")
    owner, name = m.group(1), m.group(2)
    branch = m.group(3) if m.lastindex and m.lastindex >= 3 else ""
    return owner, name, branch or ""

def _scrub(text: str, token: str) -> str:
    return text.replace(token, "***") if token else text

def shallow_clone(owner, name, branch, token, workdir):
    """Shallow, single-branch clone into a temp dir (Team C's worker does this in the real system)."""
    dest = os.path.join(workdir, "repo")
    url = f"https://github.com/{owner}/{name}.git"
    if token:
        url = f"https://x-access-token:{token}@github.com/{owner}/{name}.git"
    cmd = ["git", "clone", "--depth", "1", "--single-branch", "--no-tags"]
    if branch:
        cmd += ["--branch", branch]
    env = {**os.environ, "GIT_TERMINAL_PROMPT": "0"}
    try:
        proc = subprocess.run(cmd + [url, dest], capture_output=True, text=True, timeout=300, env=env)
    except subprocess.TimeoutExpired:
        raise RepoError("Cloning timed out after 5 minutes (repository too large or network problem).")
    if proc.returncode != 0:
        err = _scrub(proc.stderr, token).lower()
        if branch and ("remote branch" in err or "not found in upstream" in err):
            raise RepoError(f"Branch '{branch}' was not found in {owner}/{name}.")
        raise RepoError(f"Could not clone {owner}/{name}: repository not found, or it is private. "
                        "For a private repo, rerun and paste a read-only token when asked.")
    def git(*args):
        return subprocess.run(["git", "-C", dest, *args], capture_output=True, text=True).stdout.strip()
    return dest, git("rev-parse", "--short", "HEAD"), git("rev-parse", "--abbrev-ref", "HEAD")


def scan_repo(repo_dir: str):
    """Walk the clone; returns every file path (relative, posix) incl. sensitive ones (they stay visible as nodes)."""
    files, truncated = [], False
    for root, dirs, fnames in os.walk(repo_dir):
        dirs[:] = [d for d in dirs if d not in SKIP_DIRS]
        for fn in fnames:
            full = os.path.join(root, fn)
            if os.path.islink(full):
                continue
            files.append(os.path.relpath(full, repo_dir).replace(os.sep, "/"))
            if len(files) >= MAX_WALK_FILES:
                truncated = True
                break
        if truncated:
            break
    files.sort(key=lambda p: (p.count("/"), p))
    for rel in files:
        if is_sensitive_name(rel):
            JOB.excluded[rel] = "sensitive file name"
    return files, truncated
```

## S1-6 - Tech-stack detector
1. Find manifests -> 2. parse them **deterministically** -> 3. LLM classifies / ranks (structured output) -> 4. ground + validate.
The LLM may only name a framework if it points at a real dependency or file as evidence, and may only list libraries that exist in the manifests. No manifest = low-confidence fallback from file-extension statistics (no LLM call, no guessing).

```python

def _req_name(line: str):
    m = re.match(r"^\s*([A-Za-z0-9][A-Za-z0-9._\-]*)", line)
    return m.group(1) if m else None

def parse_package_json(text):
    d = json.loads(text)
    out = []
    for key in ("dependencies", "devDependencies", "peerDependencies"):
        out += [(n, "npm") for n in (d.get(key) or {})]
    return out

def parse_requirements(text):
    out = []
    for line in text.splitlines():
        line = line.split("#")[0].strip()
        if not line or line.startswith(("-", "git+", "http")):
            continue
        n = _req_name(line)
        if n:
            out.append((n, "pypi"))
    return out

def parse_pyproject(text):
    if tomllib is None:
        return []
    d = tomllib.loads(text)
    out = []
    proj = d.get("project", {})
    for s in proj.get("dependencies", []):
        n = _req_name(s)
        out += [(n, "pypi")] if n else []
    for group in proj.get("optional-dependencies", {}).values():
        out += [(_req_name(s), "pypi") for s in group if _req_name(s)]
    poetry = d.get("tool", {}).get("poetry", {})
    out += [(n, "pypi") for n in poetry.get("dependencies", {}) if n.lower() != "python"]
    for g in poetry.get("group", {}).values():
        out += [(n, "pypi") for n in g.get("dependencies", {})]
    return out

def parse_pipfile(text):
    if tomllib is None:
        return []
    d = tomllib.loads(text)
    return [(n, "pypi") for sec in ("packages", "dev-packages") for n in d.get(sec, {})]

def parse_pom(text):
    if "<!ENTITY" in text:
        raise ValueError("DTD entities not allowed")
    root = ET.fromstring(text)
    local = lambda tag: tag.rsplit("}", 1)[-1]
    out = []
    for el in root.iter():
        if local(el.tag) in ("dependency", "parent", "plugin"):
            g = a = None
            for ch in el:
                if local(ch.tag) == "groupId":
                    g = (ch.text or "").strip()
                elif local(ch.tag) == "artifactId":
                    a = (ch.text or "").strip()
            if a:
                out.append((f"{g}:{a}" if g else a, "maven"))
    return out

_GRADLE_DEP = re.compile(r"(?:implementation|api|compileOnly|runtimeOnly|testImplementation|annotationProcessor|kapt|classpath)"
                         r"\s*\(?\s*['\"]([^'\"]+)['\"]")
_GRADLE_PLUGIN = re.compile(r"\bid\s*\(?\s*['\"]([\w.\-]+)['\"]")
def parse_gradle(text):
    out = []
    for coord in _GRADLE_DEP.findall(text):
        parts = coord.split(":")
        if len(parts) >= 2:
            out.append((":".join(parts[:2]), "maven"))
    out += [(p, "gradle-plugin") for p in _GRADLE_PLUGIN.findall(text)]
    return out

def parse_gemfile(text):
    return [(n, "rubygems") for n in re.findall(r"^\s*gem\s+['\"]([^'\"]+)['\"]", text, re.M)]

def parse_gomod(text):
    return [(n, "go") for n in re.findall(r"^\s*(?:require\s+)?([A-Za-z0-9.\-_]+\.[A-Za-z]{2,}/\S+)\s+v\d", text, re.M)]

def parse_cargo(text):
    if tomllib is None:
        return []
    d = tomllib.loads(text)
    out = []
    for sec in ("dependencies", "dev-dependencies", "build-dependencies"):
        out += [(n, "crates") for n in d.get(sec, {})]
    out += [(n, "crates") for n in d.get("workspace", {}).get("dependencies", {})]
    return out

def parse_composer(text):
    d = json.loads(text)
    out = []
    for key in ("require", "require-dev"):
        out += [(n, "packagist") for n in (d.get(key) or {}) if n != "php" and not n.startswith("ext-")]
    return out

def parse_dockerfile(text):
    imgs = re.findall(r"^\s*FROM\s+(?:--\S+\s+)*([^\s]+)", text, re.M | re.I)
    return [(i, "docker") for i in imgs if i.lower() != "scratch" and "$" not in i]

MANIFEST_PARSERS = {
    "package.json": parse_package_json, "pyproject.toml": parse_pyproject, "Pipfile": parse_pipfile,
    "pom.xml": parse_pom, "build.gradle": parse_gradle, "build.gradle.kts": parse_gradle,
    "Gemfile": parse_gemfile, "go.mod": parse_gomod, "Cargo.toml": parse_cargo,
    "composer.json": parse_composer, "Dockerfile": parse_dockerfile,
}
REQ_RE = re.compile(r"^requirements.*\.txt$", re.I)
MANAGER_BY_FILE = {
    "package.json": "npm", "package-lock.json": "npm", "yarn.lock": "yarn", "pnpm-lock.yaml": "pnpm",
    "bun.lockb": "bun", "pipfile": "pipenv", "pipfile.lock": "pipenv", "poetry.lock": "poetry", "uv.lock": "uv",
    "pom.xml": "maven", "build.gradle": "gradle", "build.gradle.kts": "gradle", "gemfile": "bundler",
    "go.mod": "go modules", "cargo.toml": "cargo", "composer.json": "composer",
}


def language_stats(repo_dir, files):
    sizes = Counter()
    for rel in files:
        if rel in JOB.excluded:
            continue
        p = PurePosixPath(rel)
        lang = EXT_LANG.get(p.suffix.lower())
        low = p.name.lower()
        if not lang or low in LOCK_FILES or low.endswith((".min.js", ".min.css")):
            continue
        try:
            sizes[lang] += os.path.getsize(os.path.join(repo_dir, rel))
        except OSError:
            pass
    total = sum(sizes.values()) or 1
    return [LanguageShare(name=n, share=round(s / total, 2)) for n, s in sizes.most_common(8) if s / total >= 0.01]


def find_entry_candidates(repo_dir, files, max_items=8):
    NOISE = {"tests", "test", "docs", "doc", "examples", "example", "e2e", "fixtures"}
    cands = [f for f in files if f not in JOB.excluded and f.count("/") <= 3
             and not (set(PurePosixPath(f).parts[:-1]) & NOISE)
             and PurePosixPath(f).name.lower() in ENTRY_NAMES]
    if "package.json" in files:
        txt = safe_read_text(repo_dir, "package.json")
        try:
            main = json.loads(txt).get("main") if txt else None
            main = main.removeprefix("./") if isinstance(main, str) else None
            if main and main in files and main not in cands:
                cands.insert(0, main)
        except Exception:
            pass
    return cands[:max_items]


STACK_SYSTEM = ("You are a precise software analyst. Use ONLY the facts you are given. Never invent dependencies, "
                "files or technologies. Respond only with JSON that matches the required schema.")
STACK_USER = """Below are the dependencies extracted from the manifest files of a repository, plus a sample of its file names.

MANIFEST FILES: {manifests}
DETECTED LANGUAGES: {languages}
DEPENDENCIES (name [ecosystem]):
{dependencies}
SAMPLE FILE NAMES: {files}

Task:
1. frameworks: the application frameworks / platforms this project is built with (for example a web framework, UI framework, ORM or test framework). For each give "name" and "evidence". The evidence MUST be copied exactly from one dependency name in the list above (name only, WITHOUT the [ecosystem] tag) or an exact sample file name that proves the framework.
2. primary_libraries: up to 10 of the most important libraries that are NOT frameworks. Each MUST be copied exactly from the dependency list.
Rank by importance, most important first. Leave out anything the lists above do not support."""

stack_chain = StructuredChain("tech-stack", STACK_SYSTEM, STACK_USER, StackDraft)


def detect_tech_stack(repo_dir, files):
    languages = language_stats(repo_dir, files)

    deps, managers, manifests_found = {}, [], []
    for rel in files:
        if rel in JOB.excluded or rel.count("/") > MANIFEST_MAX_DEPTH:
            continue
        name = PurePosixPath(rel).name
        parser = MANIFEST_PARSERS.get(name) or (parse_requirements if REQ_RE.match(name) else None)
        mgr = MANAGER_BY_FILE.get(name.lower()) or ("pip" if REQ_RE.match(name) else None)
        if mgr and mgr not in managers:
            managers.append(mgr)
        if not parser:
            continue
        text = safe_read_text(repo_dir, rel)
        if text is None:
            continue
        manifests_found.append(rel)
        if name == "pyproject.toml":
            mgr2 = "poetry" if "[tool.poetry" in text else "pip"
            if mgr2 not in managers:
                managers.append(mgr2)
        try:
            for dep, eco in parser(text):
                deps.setdefault(dep.lower(), (dep, eco, rel))
        except Exception as e:
            JOB.warnings.append(f"could not parse {rel}: {type(e).__name__}")
    if "npm" in managers and "package-lock.json" not in files and any(m in managers for m in ("yarn", "pnpm", "bun")):
        managers.remove("npm")

    meta = {"manifests": manifests_found, "dependency_count": len(deps)}

    if not deps:
        JOB.warnings.append("No manifest/dependency files found: tech stack is a low-confidence fallback "
                            "based on file-extension statistics only.")
        return TechStack(languages=languages, frameworks=[], libraries=[], package_managers=managers,
                         confidence="low"), meta

    dep_lines = [f"- {d} [{eco}]" for d, eco, _ in list(deps.values())[:200]]
    sample = [PurePosixPath(f).name for f in files if f not in JOB.excluded][:40]
    try:
        draft = stack_chain(manifests=", ".join(manifests_found[:15]),
                            languages=", ".join(f"{l.name} {int(l.share*100)}%" for l in languages) or "unknown",
                            dependencies="\n".join(dep_lines), files=", ".join(sample))
        log("llm", f"draft: {draft}")
    except (SchemaValidationError, ContextOverflow, RuntimeError) as e:
        JOB.warnings.append(f"LLM classification failed ({e}); frameworks omitted, confidence lowered.")
        return TechStack(languages=languages, frameworks=[], libraries=[], package_managers=managers,
                         confidence="low"), meta

    file_names = {PurePosixPath(f).name.lower() for f in files if f not in JOB.excluded}
    frameworks, libraries, dropped = [], [], 0
    _tag = re.compile(r"\s*\[[^\]]*\]\s*$")
    for g in draft.frameworks:
        ev, nm = _tag.sub("", g.evidence.strip()).lower(), g.name.strip()
        if nm and (ev in deps or ev in file_names):
            if nm.lower() not in {x.lower() for x in frameworks}:
                frameworks.append(nm)
        else:
            dropped += 1
    for lib in draft.primary_libraries:
        key = _tag.sub("", lib.strip()).lower()
        if key in deps:
            orig = deps[key][0]
            if orig not in libraries and len(libraries) < 10:
                libraries.append(orig)
        else:
            dropped += 1
    if dropped:
        JOB.warnings.append(f"{dropped} LLM suggestion(s) were dropped because no manifest evidence backed them.")
    if not libraries:
        libraries = [d for d, eco, rel in deps.values() if rel.count("/") == 0 and eco != "docker"][:10]
        if libraries:
            JOB.warnings.append("LLM returned no libraries; used root-manifest dependencies instead.")
    confidence = "high" if dropped == 0 and (frameworks or libraries) else "medium"
    return TechStack(languages=languages, frameworks=frameworks, libraries=libraries,
                     package_managers=managers, confidence=confidence), meta
```

## S1-7 - Chunking + embedding scaffold (in-memory, retrieval-free)
Load -> language-aware split -> attach `file_path / start_line / end_line / language / repo_id / branch / tenant_id` -> batch-embed with backoff -> store in an **in-memory FAISS index** (lives only in this notebook's RAM; nothing is written to disk or a server). Chunk IDs are derived from repo + path + content hash so re-indexing is repeatable.

```python
def get_splitter(lang_name, cache={}):
    if lang_name in cache:
        return cache[lang_name]
    enum_name = {"Python": "PYTHON", "JavaScript": "JS", "TypeScript": "TS", "Java": "JAVA", "Kotlin": "KOTLIN",
                 "Go": "GO", "Rust": "RUST", "Ruby": "RUBY", "PHP": "PHP", "C#": "CSHARP", "C": "C", "C++": "CPP",
                 "Swift": "SWIFT", "Scala": "SCALA", "HTML": "HTML", "Markdown": "MARKDOWN"}.get(lang_name)
    lang = getattr(Language, enum_name, None) if enum_name else None
    kwargs = dict(chunk_size=CHUNK_SIZE, chunk_overlap=CHUNK_OVERLAP, add_start_index=True)
    cache[lang_name] = (RecursiveCharacterTextSplitter.from_language(language=lang, **kwargs) if lang
                        else RecursiveCharacterTextSplitter(**kwargs))
    return cache[lang_name]


def load_and_chunk(repo_dir, files, repo_id, branch):
    """Eligible, non-sensitive files -> chunk Documents with citation metadata. Shallow files first (cost cap)."""
    docs, file_lines, truncated = [], {}, False
    for rel in files:
        if not is_indexable(rel) or rel in JOB.excluded:
            continue
        if len(docs) >= MAX_CHUNKS:
            truncated = True
            break
        text = safe_read_text(repo_dir, rel)
        if text is None or not text.strip():
            continue
        file_lines[rel] = text.count("\n") + 1
        lang = language_of(rel)
        meta = {"file_path": rel, "language": lang, "repo_id": repo_id, "branch": branch, "tenant_id": TENANT_ID}
        for c in get_splitter(lang).create_documents([text], metadatas=[meta]):
            si = max(c.metadata.get("start_index", 0), 0)
            start_line = text.count("\n", 0, si) + 1
            end_line = start_line + c.page_content.rstrip("\n").count("\n")
            c.metadata.update(start_line=start_line, end_line=min(end_line, file_lines[rel]))
            c.metadata["chunk_id"] = hashlib.sha1(
                f"{repo_id}|{rel}|{hashlib.sha1(c.page_content.encode()).hexdigest()}|{start_line}".encode()).hexdigest()
            docs.append(c)
    return docs[:MAX_CHUNKS], file_lines, truncated


def assert_no_sensitive(docs):
    """Guard check: nothing excluded (by name or by content) may be in the chunks that get embedded."""
    for d in docs:
        p = d.metadata["file_path"]
        if p in JOB.excluded or is_sensitive_name(p) or has_secret_content(d.page_content):
            raise RuntimeError(f"GUARD VIOLATION: sensitive content reached the embedding step ({p})")
    return len(docs)


def get_embedder():
    """Provider-agnostic adapter point: swap this function to change the embedding provider (NFR-09)."""
    import torch
    device = "cuda" if torch.cuda.is_available() else "cpu"
    try:
        from langchain_huggingface import HuggingFaceEmbeddings as _HFE
    except ImportError:
        _HFE = HuggingFaceEmbeddings
    emb = _HFE(model_name=EMBED_MODEL, model_kwargs={"device": device},
               encode_kwargs={"batch_size": EMBED_BATCH, "normalize_embeddings": True})
    return emb, device


def embed_with_backoff(embedder, texts):
    vectors = []
    for i in range(0, len(texts), EMBED_BATCH):
        batch = texts[i:i + EMBED_BATCH]
        for attempt in range(MAX_RETRIES):
            try:
                vectors.extend(embedder.embed_documents(batch))
                break
            except Exception as e:
                if attempt == MAX_RETRIES - 1:
                    raise
                wait = 2 ** attempt
                log("embed", f"batch {i//EMBED_BATCH} failed ({type(e).__name__}); backing off {wait}s")
                time.sleep(wait)
    return vectors


def build_index(docs):
    embedder, device = get_embedder()
    texts = [d.page_content for d in docs]
    vectors = embed_with_backoff(embedder, texts)
    store = FAISS.from_embeddings(list(zip(texts, vectors)), embedding=embedder,
                                  metadatas=[d.metadata for d in docs],
                                  ids=[d.metadata["chunk_id"] for d in docs])
    JOB.cost["embed_chunks"] += len(docs)
    JOB.cost["embed_tokens_est"] += sum(len(t) for t in texts) // 4
    return store, device
```

## S1-8 in action - Module-map chain
The model sees a compact, secret-free summary of the repository (directory counts, file list with line counts, entry-point candidates, README excerpt, first lines of key files) and must return a `ModuleMap`. Afterwards every module is **grounded**: each `SourceRef` must point at a real indexed file with an in-range line span, otherwise the reference is dropped, and a module with no verifiable source is removed and flagged.

```python
MODULE_SYSTEM = ("You are a senior engineer documenting a codebase for new teammates. Write in plain, non-jargon "
                 "language. Use ONLY files and paths that appear in the information you are given. "
                 "Respond only with JSON that matches the required schema.")
MODULE_USER = """Repository: {repo_name}
Detected stack: {stack_summary}

README excerpt:
{readme}

DIRECTORY SUMMARY (path: number of files):
{dirs}

FILES (path, total lines):
{files}

ENTRY POINT CANDIDATES:
{entry_candidates}

KEY FILE HEADS (line number | code):
{heads}

Task - describe the codebase as a module map:
- modules: between 3 and 8 modules. Each has "name", "summary" (one or two plain sentences), "paths" (directories or files copied exactly from the lists above) and "sources" (1 to 3 references, each with "file_path" copied exactly from the FILES list and "start_line"/"end_line" inside that file's line count).
- entry_points: chosen only from the ENTRY POINT CANDIDATES (empty list if none fit).
- dependencies: pairs with keys "from" and "to", each a module name defined above, meaning that module "from" uses module "to"."""

module_chain = StructuredChain("module-map", MODULE_SYSTEM, MODULE_USER, ModuleMapDraft)


def build_module_inputs(repo_dir, repo_id, stack, file_lines, entry_cands, budget):
    n_files, head_lines, readme_chars = 150, 40, 2000
    while True:
        counts = Counter()
        for p in file_lines:
            parts = p.split("/")
            counts["(repo root)" if len(parts) == 1 else parts[0]] += 1
            if len(parts) > 2:
                counts["/".join(parts[:2])] += 1
        dirs = "\n".join(f"{d}: {n}" for d, n in sorted(counts.items())[:40])

        LOW = {"docs", "doc", "tests", "test", "examples", "example", ".github", ".devcontainer"}
        shown = sorted(file_lines, key=lambda p: (("/" in p and p.split("/")[0].lower() in LOW),
                                                  p.count("/"), p))[:n_files]
        files_txt = "\n".join(f"{p} ({file_lines[p]} lines)" for p in shown)

        readme_rel = next((p for p in sorted(file_lines, key=lambda p: (p.count("/"), p))
                           if PurePosixPath(p).name.lower().startswith("readme")), None)
        readme = (safe_read_text(repo_dir, readme_rel) or "")[:readme_chars] if readme_rel else ""

        head_files = entry_cands[:4] or [p for p in shown if language_of(p) in EXT_LANG.values()][:3]
        heads = []
        for p in head_files:
            txt = safe_read_text(repo_dir, p)
            if txt:
                lines = txt.splitlines()[:head_lines]
                heads.append(f"### {p}\n" + "\n".join(f"{i:>3}| {l[:120]}" for i, l in enumerate(lines, 1)))

        inputs = dict(
            repo_name=repo_id,
            stack_summary=("languages: " + ", ".join(f"{l.name} {int(l.share*100)}%" for l in stack.languages)
                           + "; frameworks: " + (", ".join(stack.frameworks) or "unknown")
                           + "; package managers: " + (", ".join(stack.package_managers) or "unknown")),
            readme=readme or "(no README found)",
            dirs=dirs or "(none)",
            files=files_txt or "(none)",
            entry_candidates="\n".join(entry_cands) or "(none)",
            heads="\n\n".join(heads) or "(none)",
        )
        if sum(len(v) for v in inputs.values()) <= budget or (n_files <= 20 and head_lines <= 15):
            return inputs
        n_files, head_lines, readme_chars = (max(20, int(n_files * 0.7)), max(15, head_lines - 8),
                                             max(400, int(readme_chars * 0.7)))


def ground_module_map(draft, file_lines, entry_cands):
    """Keep only what can be verified against the real repository."""
    warnings_, modules, seen = [], [], set()
    dir_set = {"/".join(p.split("/")[:i]) for p in file_lines for i in range(1, len(p.split("/")))}
    for m in draft.modules:
        refs = []
        for s in m.sources:
            path = s.file_path.strip().removeprefix("./")
            n = file_lines.get(path)
            if n is None:
                continue
            start = max(1, min(s.start_line, n))
            end = max(start, min(s.end_line, n))
            refs.append(SourceRef(file_path=path, start_line=start, end_line=end))
        if not refs:
            warnings_.append(f"Module '{m.name}' removed: no verifiable source reference (file/lines).")
            continue
        if m.name.strip().lower() in seen or not m.name.strip():
            continue
        seen.add(m.name.strip().lower())
        paths = []
        for p in m.paths:
            q = p.strip().removeprefix("./").rstrip("/")
            if q in file_lines:
                paths.append(q)
            elif q in dir_set:
                paths.append(q + "/")
        modules.append(Module(name=m.name.strip(), summary=m.summary.strip(), paths=paths, sources=refs))
    names = {m.name for m in modules}
    deps, seen_d = [], set()
    for d in draft.dependencies:
        if d.from_ in names and d.to in names and d.from_ != d.to and (d.from_, d.to) not in seen_d:
            seen_d.add((d.from_, d.to))
            deps.append(ModuleDependency(from_=d.from_, to=d.to))
    entries = [e.strip().removeprefix("./") for e in draft.entry_points if e.strip().removeprefix("./") in file_lines]
    if not entries:
        entries = [e for e in entry_cands if e in file_lines]
    return ModuleMap(modules=modules, entry_points=entries, dependencies=deps), warnings_
```

## Pipeline orchestration and the terminal report

```python
def analyze_repo(repo_input, branch_input="", token="", quick=False):
    global JOB
    JOB = Job()
    owner, name, branch_hint = parse_repo_url(repo_input)
    repo_id = f"{owner}/{name}"
    workdir = tempfile.mkdtemp(prefix="onboarding_")
    result = {"repo_id": repo_id, "errors": []}
    log("job", f"analysing {repo_id}")
    try:
        with stage("clone"):
            repo_dir, commit, branch = shallow_clone(owner, name, branch_input or branch_hint, token, workdir)
            result.update(branch=branch, commit=commit)

        with stage("scan"):
            files, truncated = scan_repo(repo_dir)
            result["files"] = files
            if truncated:
                JOB.warnings.append(f"Repository has more than {MAX_WALK_FILES} files; scan was truncated.")
            log("scan", f"{len(files)} files, {len(JOB.excluded)} excluded by name before any AI step")

        with stage("tech-stack"):
            result["entry_candidates"] = find_entry_candidates(repo_dir, files)
            stack, meta = detect_tech_stack(repo_dir, files)
            result.update(tech_stack=stack, stack_meta=meta)

        with stage("index"):
            docs, file_lines, cap_hit = load_and_chunk(repo_dir, files, repo_id, branch)
            if cap_hit:
                JOB.warnings.append(f"Chunk cap ({MAX_CHUNKS}) reached: deeper files were not indexed.")
            checked = assert_no_sensitive(docs)
            result["file_lines"] = file_lines
            if docs:
                store, device = build_index(docs)
            else:
                store, device = None, "n/a"
                JOB.warnings.append("No indexable text files were found.")
            result["vectorstore"] = store
            result["index_status"] = {"files_indexed": len(file_lines), "chunks": len(docs), "guard_checked": checked,
                                      "embed_model": EMBED_MODEL, "device": device, "truncated": cap_hit,
                                      "est_tokens": JOB.cost["embed_tokens_est"]}

        with stage("module-map"):
            result["module_map"], result["module_warnings"] = None, []
            if quick:
                JOB.warnings.append("Module map skipped (quick mode).")
            elif file_lines:
                for budget in (14_000, 9_000, 5_000):
                    try:
                        inputs = build_module_inputs(repo_dir, repo_id, stack, file_lines,
                                                     result["entry_candidates"], budget)
                        draft = module_chain(**inputs)
                        mm, warns = ground_module_map(draft, file_lines, result["entry_candidates"])
                        result["module_map"], result["module_warnings"] = mm, warns
                        break
                    except ContextOverflow as e:
                        log("module-map", f"prompt too large, shrinking ({e})")
                    except (SchemaValidationError, RuntimeError) as e:
                        result["errors"].append(f"module-map: {e}")
                        break
    finally:
        shutil.rmtree(workdir, ignore_errors=True)
        log("cleanup", "temporary clone deleted; only derived data remains in memory")
    result["excluded"] = dict(JOB.excluded)
    result["job"] = JOB
    return result


def print_report(r):
    W = 78
    job = r["job"]
    def title(t):
        print("\n" + "=" * W + f"\n {t}\n" + "=" * W)

    title("REPOSITORY")
    print(f" Repo        : {r['repo_id']}")
    print(f" Branch      : {r['branch']}   (commit {r['commit']})")
    print(f" Files found : {len(r['files'])}   | content excluded (secrets): {len(r['excluded'])}"
          f"   | skipped: {dict(job.skipped) or 'none'}")

    title("CODEBASE MAP (top levels)")
    files = r["files"]
    dir_counts, root_files = Counter(), []
    for p in files:
        parts = p.split("/")
        if len(parts) == 1:
            root_files.append(p)
        else:
            dir_counts[parts[0]] += 1
            if len(parts) > 2:
                dir_counts[parts[0] + "/" + parts[1]] += 1
    tops = sorted(d for d in dir_counts if "/" not in d)
    print(f" {r['repo_id'].split('/')[1]}/")
    for d in tops[:25]:
        print(f"   {d + '/':<30} ({dir_counts[d]} files)")
        subs = sorted(s for s in dir_counts if s.startswith(d + "/"))
        for s in subs[:8]:
            print(f"       {s.split('/')[1] + '/':<26} ({dir_counts[s]} files)")
        if len(subs) > 8:
            print(f"       ... {len(subs) - 8} more")
    if len(tops) > 25:
        print(f"   ... {len(tops) - 25} more top-level folders")
    for f in root_files[:15]:
        tag = "  [EXCLUDED]" if f in r["excluded"] else ""
        print(f"   {f}{tag}")
    if len(root_files) > 15:
        print(f"   ... {len(root_files) - 15} more root files")
    print("\n Entry-point candidates:", ", ".join(r["entry_candidates"]) or "none found")
    if r["excluded"]:
        print(" Excluded from every AI step (content never read):")
        for p, why in list(r["excluded"].items())[:15]:
            print(f"   [EXCLUDED] {p}  - {why}")
        if len(r["excluded"]) > 15:
            print(f"   ... {len(r['excluded']) - 15} more")

    ts = r["tech_stack"]
    title(f"TECH STACK   (confidence: {ts.confidence.upper()})")
    for l in ts.languages:
        print(f"   {l.name:<18} {'#' * int(l.share * 30):<30} {int(l.share * 100):>3}%")
    print(f" Frameworks       : {', '.join(ts.frameworks) or '-'}")
    print(f" Libraries        : {', '.join(ts.libraries) or '-'}")
    print(f" Package managers : {', '.join(ts.package_managers) or '-'}")
    print(f" Manifests parsed : {', '.join(r['stack_meta']['manifests'][:8]) or 'none'}"
          f"  ({r['stack_meta']['dependency_count']} dependencies)")

    mm = r["module_map"]
    title("MODULE MAP")
    if mm is None:
        print(" Not available:", "; ".join(r["errors"]) or "skipped (quick mode) or no indexable files")
    else:
        for i, m in enumerate(mm.modules, 1):
            print(f" {i}. {m.name}")
            print(textwrap.fill(m.summary, 74, initial_indent="    ", subsequent_indent="    "))
            print(f"    paths  : {', '.join(m.paths) or '-'}")
            print(f"    source : " + ", ".join(f"{s.file_path}:{s.start_line}-{s.end_line}" for s in m.sources))
        print("\n Entry points :", ", ".join(mm.entry_points) or "-")
        print(" Dependencies :", "; ".join(f"{d.from_} -> {d.to}" for d in mm.dependencies) or "-")
        for w in r["module_warnings"]:
            print(" [flag]", w)

    ix = r["index_status"]
    title("INDEX STATUS (in-memory FAISS, not persisted)")
    print(f" Files indexed : {ix['files_indexed']}   Chunks: {ix['chunks']}   (~{ix['est_tokens']:,} tokens)")
    print(f" Embedding     : {ix['embed_model']} on {ix['device']}")
    print(f" Metadata      : file_path, start_line, end_line, language, repo_id, branch, tenant_id on every chunk")

    title("SAFETY, COST & TIMINGS")
    print(f" Guard check   : {ix['guard_checked']} chunks verified free of excluded/secret content")
    print(f" LLM           : {job.cost['llm_calls']} calls, {job.cost['llm_prompt_tokens']:,} prompt tokens, "
          f"{job.cost['llm_output_tokens']:,} output tokens")
    print(f" Stage timings : " + ", ".join(f"{k} {v:.1f}s" for k, v in job.timings.items()))
    for w in job.warnings:
        print(" [warning]", w)
    for e in r["errors"]:
        print(" [error]  ", e)

    title("JSON CONTRACTS (what Teams A and C would receive)")
    print("TechStack =", json.dumps(ts.model_dump(), indent=2))
    if mm is not None:
        print("ModuleMap =", json.dumps(mm.model_dump(by_alias=True), indent=2))
```
